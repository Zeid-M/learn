```python
import os
import smtplib
from email.mime.application import MIMEApplication
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
```

#### Define the host, port, email, and password

```python
SMTP_HOST = ""
SMTP_PORT = "" # "587" for TTL or "465" for SSL
EMAIL = ""
EMAIL_PASSWORD = ""
```

#### Define the subject, content, attachments, to, cc, bcc

```python
subject=""
content=""
attachments=[]
to=[]
cc=[]
bcc=[]
```

#### SMTP session for TTL

```python
# Start the SMTP session and login to your account
smtp_session = smtplib.SMTP(SMTP_HOST, SMTP_PORT)
smtp_session.starttls()
smtp_session.login(EMAIL, EMAIL_PASSWORD)
```

#### Or

#### SMTP session for SSL

```python
# Start the SMTP session and login to your account
smtp_session = smtplib.SMTP_SSL(SMTP_HOST, SMTP_PORT)
smtp_session.login(EMAIL, EMAIL_PASSWORD)
```

#### Setup the message

```python
msg = MIMEMultipart()
msg["From"] = EMAIL
msg["To"] = ",".join(to)
msg["Subject"] = subject
msg["CC"] = ",".join(cc)
msg["BCC"] = ",".join(bcc)
```

#### Setup the attachments

```python
msg.attach(MIMEText(content, "plain"))

if attachments:
    for file_path in attachments or []:
        with open(file_path, "rb") as fil:
            part = MIMEApplication(fil.read(), Name=os.path.basename(file_path))
        # After the file is closed
        part["Content-Disposition"] = (
            f'attachment; filename="{os.path.basename(file_path)}"'
        )
        msg.attach(part)
```

#### Send the email

```python
    # Send the email
print(msg.as_string())
smtp_session.sendmail(EMAIL, [*to, *cc, *bcc], msg.as_string())
```

#### Close the SMTP session

```python
smtp_session.close()
```
