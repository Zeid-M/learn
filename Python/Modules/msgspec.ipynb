##### Define your message schemas using standard Python type annotations.

```python
import msgspec

class User(msgspec.Struct):
    """A new type describing a User"""
    name: str # required
    groups: set[str] = set() # optional
    email: str | None = None # optional
```

```python
alice = User("alice", groups={"admin", "engineering"})

alice
```

```python
alice.name
```

##### Encode messages as JSON

```python
msg = msgspec.json.encode(alice)

msg
```

##### Decode messages back into Python objects
- It will only take the defined argument in the class

```python
msg='{"name":"Zeid","groups":["engineering","trainer"],"email":"zeid@gmail.com","extra":"..."}'
```

```python
zeid= msgspec.json.decode(msg, type=User)

zeid
```

##### optional schema validation

```python
msgspec.json.decode(b'{"name":"bob","groups":[123]}', type=User)
```

```python
msgspec.json.decode(b'{"groups":["admin"]}', type=User)
```
