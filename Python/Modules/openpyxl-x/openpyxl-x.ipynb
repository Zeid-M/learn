```python
import openpyxl as xl
```

#### Open the workbook

```python
wb = xl.load_workbook("files.xlsx")
```

- select the sheet

```python
sheet = wb["Sheet1"]
```

- change the column dimensions

```python
sheet.column_dimensions["A"].width = 2
sheet.column_dimensions["B"].width = 6
sheet.column_dimensions["C"].width = 4
sheet.column_dimensions["D"].width = 8
```

- save the file

```python
wb.save("files_1.xlsx")
```
