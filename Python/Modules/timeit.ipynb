```python
import timeit
```

```python
mysetup = ''' 
import msgspec

class User(msgspec.Struct):

    name: str # required
    groups: set[str] = set() # optional
    email: str | None = None # optional

json={"name":"Zeid","groups":["engineering","trainer"],"email":"zeid@gmail.com","extra":"..."}

msg = msgspec.json.encode(json)

zeid= msgspec.json.decode(msg, type=User)


'''
 
mycode = ''' 
zeid.email
'''
 
# timeit statement
print(timeit.timeit(setup=mysetup,
                    stmt=mycode,
                    number=10000))
```

```python

mysetup = ''' 
json={"name":"Zeid","groups":["engineering","trainer"],"email":"zeid@gmail.com","extra":"..."}
'''
 

mycode = ''' 
json["email"]
'''
 

print(timeit.timeit(setup=mysetup,
                    stmt=mycode,
                    number=10000))
```

```python
 
# importing the required module
import timeit
 
# code snippet to be executed only once
mysetup = "from math import sqrt"
 
# code snippet whose execution time is to be measured
mycode = ''' 
def example(): 
    mylist = [] 
    for x in range(100): 
        mylist.append(sqrt(x)) 
'''
 
# timeit statement
print(timeit.timeit(setup=mysetup,
                    stmt=mycode,
                    number=10000))
```
