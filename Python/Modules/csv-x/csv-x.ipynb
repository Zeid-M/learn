```python
import csv
```

### Read a csv file:
#### Open the csv file

```python
file = open("file_1.csv")
```

#### Use the `csv.reader` object to read the csv file

```python
csvreader = csv.reader(file)
```

#### Extract the field names

```python
header = []
header = next(csvreader)
```

```python
header
```

#### Extract the rows

```python
rows = []
for row in csvreader:
    rows.append(row)
```

```python
rows
```

#### Close the file

```python
file.close()
```

### Write a csv file:
#### define a filename and open the file

```python
filename = "file_2.csv"
header=["Name","Age"]
data=[["Zeid",24]]
```

#### to avoid forgeten to close the file use with() statment

```python
with open(filename, "w", newline="") as file:
    # create a csvwriter object
    csvwriter = csv.writer(file)

    # write the header (single line)
    csvwriter.writerow(header)

    # write the rest of the data (multi line)
    csvwriter.writerows(data)
```
