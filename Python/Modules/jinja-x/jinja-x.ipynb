```python
from jinja2 import Environment, FileSystemLoader
```

```python
templates_path="./templates"
template_name="helloName.jinja"
```

#### Create a Jinja2 environment using the directory of the templates

```python
env = Environment(loader=FileSystemLoader(templates_path))
```

#### Load the template from the filename

```python
template = env.get_template(template_name)
```

#### Render the email template with the arguments

```python
body = template.render(name="Zeid")
```

#### Return the rendered email body

```python
body
```

#### Rendere a template with logic in and save the output to a files

```python
template = env.get_template("webTable.jinja")

with open("renders/output.html", "w") as f:
    print(template.render(tableOf=5), file=f)
```
