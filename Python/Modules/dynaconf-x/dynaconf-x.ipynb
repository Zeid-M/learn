```python
from dynaconf import Dynaconf, Validator
```

#### Setup the dynaconf object

```python
settings = Dynaconf(
    settings_files=[
        "/home/zeid_m/Code/GitHub/learn/Python/Modules/dynaconf/db_config.toml"
    ],  # files to load the settings from(toml,yaml,json,.py)
    envvar_prefix="DYNACONF",  # will get an environments variables that start with the prefix
    environments=True,  # enable environments (testing, production, ...)
    env_switcher="ENV_FOR_DYNACONF",  # you can switch the env by setting this environment variable
    default_env="testing",  # the default env that will fallback to when not finding the object in the current env, if not named the default env is "default"
    env="production",  # set the current env
)
```

#### Setup the settings files directory, the setting files, and any environments variables in you environment (or .env file)

```python
ROOT_PATH_FOR_DYNACONF='/path/to/config_dir'
SETTINGS_FILES_FOR_DYNACONF=["config.toml"]
DYNACONF_FOLDER='/path/to/folder'
```

#### Setup the validators

```python
settings.validators.register(
    Validator("FOLDER"), # it can exist or not
    Validator("mysql.db", must_exist=True, env="testing"), # must_exist = true means it have to exist, must_exist = false means it must not exist
    Validator("mysql.host", must_exist=True, env="testing"),
    Validator("mysql.db", must_exist=True, env="testing"),
    Validator("mysql.host", must_exist=True, env="testing"),
    Validator("mysql.root.username", must_exist=True, env="testing"),
    Validator(
        "mysql.root.password",
        must_exist=True,
        when=Validator("mysql.root.username", must_exist=True, env="testing"),
        env="testing",
    ),
    Validator("mysql.admin.username", must_exist=True, env="testing"),
    Validator(
        "mysql.admin.password",
        must_exist=True,
        when=Validator("mysql.admin.username", must_exist=True, env="testing"),
        env="testing",
    ),
    Validator("mysql.root.username", must_exist=True, env="production"),
    Validator(
        "mysql.root.password",
        must_exist=True,
        when=Validator("mysql.root.username", must_exist=True, env="production"),
        env="production",
    ),
    Validator("mysql.admin.username", env="production"),
    Validator(
        "mysql.admin.password",
        must_exist=True,
        when=Validator("mysql.admin.username", must_exist=True, env="production"),
        env="production",
    ),
)
```

#### Validate

```python
settings.validators.validate()
```

```python
settings.validators.validate_all() # will get all validation errors then print them
```

```python
print(settings.mysql.root.username)
print(settings.mysql.root.password)
print(settings.mysql.admin.username) # will get it form the default env if not found
print(settings.mysql.admin.password)
```

```python
with settings.using_env("production"): # enforce useing this evn and will not fallback to the default env
    print(settings.mysql.root.username)
    print(settings.mysql.root.password)
    print(settings.mysql.admin.username)
    print(settings.mysql.admin.password)
```
