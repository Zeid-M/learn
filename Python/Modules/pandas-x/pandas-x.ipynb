#### Dataframe manplation

- Create an empty dataframe

```python
df = pd.DataFrame()
# df = pd.DataFrame()
```

- Adding a new column

```python
df["a"] = [1, 2, 3]  # a column with values
df["b"] = ""  # a empty column
df["c"] = 0  # a column with a default value (e.g 0)
df["d"] = [4, 5, 6]  # a column with values
```

- Using the `insert()` method

```python
df.insert(
    1,  # the index of the column
    "column_name",  # name of the column
    "value",  # the value of the column
)
```

- Get a column

```python
result = df["a"]
result
```

- Get data:
    - loc: by name
    - iloc: by index

- Get a row

```python
result = df.loc[1]  # row name
result
```

```python
result = df.iloc[1]  # row index
result
```

- Get a column

```python
#              row, column
result = df.loc[:, "column_name"]
result
```

```python
#              row, column
result = df.iloc[:, 1]
result
```

- Get a value (cell)

```python
#              row, column
result = df.iloc[1, 0]
result
```

```python
#         row index, column
result = df.iloc[0]["a"]
result
```

```python
#      index name, column name
result = df.loc[2, "a"]
result
```

- Get multiple rows and columns

```python
result = df.iloc[:, [1, 3]]
result
```

```python
result = df.loc[:, ["a", "d"]]
result
```

- Get rows using Conditions

```python
result = df.loc[df["a"] == 1, "d"]
result
```

```python
result = df.loc[(df["a"] == 1) & (df["d"] != 1)]
result
```

- `get()` method

```python
result=df.get("d",default="Null")
result
```

```python
result=df.get("g",default="1")
result
```

- add a value to row

```python
df.loc[df.index[1], "b"] = "13"
df
```

- Append a row

```python
df.loc[len(df)] = [4, "value", 14, 0, 7]
df
```

- `isin()` method

```python
df["e"] = df["a"].isin([1, 3])
df
```

- Shift a column

```python
df["prev_a"] = df["a"].shift()  # shift down
df["next_a"] = df["a"].shift(-1)  # shift up
df
```

- Compare columns

```python
df["equal"] = df["a"].eq(df["d"])
df
```

- Apply a function to each row or column using `apply()`

```python
df1 = df.apply(lambda x: x * 10, axis=0)  # rows
df2 = df.apply(lambda x: x * 5, axis=1)  # columns
```

```python
df1
```

```python
df2
```

```python
def squaredata(x):
    return x * x
```

```python
df3 = df["d"].apply(squaredata)
df3
```

```python
df["new_col"] = df.apply(lambda row: row.iloc[0] + row.iloc[3] + row.iloc[4], axis=1)
df
```

- Concatenate DataFrames

```python
df4 = pd.concat([df1, df3])
df4
```

```python
df5 = pd.concat([df1, df3], axis=1)
df5
```

```python
countries = {
    1: {
        "Country": "New Country 1",
        "Capital": "New Capital 1",
        "Population": "123,456,789",
    },
    2: {
        "Country": "New Country 2",
        "Capital": "New Capital 2",
        "Population": "987,654,321",
    },
    3: {
        "Country": "New Country 3",
        "Capital": "New Capital 3",
        "Population": "111,222,333",
    },
}

df6 = pd.DataFrame.from_dict(countries, orient="index")
df6
```

- Save the dataframes into a single xlsx file with multiple sheets

```python
with pd.ExcelWriter(f"files.xlsx", mode="w") as writer:
    df.to_excel(writer, index=False, sheet_name="df")
    df1.to_excel(writer, index=False, sheet_name="df1")
    df2.to_excel(writer, index=False, sheet_name="df2")
```

```python
with pd.ExcelWriter(f"files.xlsx", mode="a") as writer: # append
    df3.to_excel(writer, index=False, sheet_name="df3")
    df4.to_excel(writer, index=False, sheet_name="df4")
    df5.to_excel(writer, index=False, sheet_name="df5")
    df6.to_excel(writer, index=False, sheet_name="df6")
```
