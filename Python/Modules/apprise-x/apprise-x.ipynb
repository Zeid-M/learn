```python
import apprise
from apprise import AppriseConfig, NotifyType
import logging
```

#### Create an AppriseConfig() object and an Apprise object

```python
config = AppriseConfig()
apobj = apprise.Apprise()
```

#### Add a configuration file by it's local path

```python
config.add("file://./.apprise.yaml")
apobj.add(config)
```

#### Or you can add a apprise instance directly

```python
apobj.add("", tag="tag")
```

#### Send notifications to all define destinations or for specific tag only

```python
apobj.notify(
    body="body",
    title="title",
    tag="tag",
    notify_type=NotifyType.WARNING,
)
```

#### You can get more logging in this way

```python
logging.basicConfig(level=logging.INFO)

logger = logging.getLogger(__name__)

with apprise.LogCapture(level=apprise.logging.DEBUG) as logs:

    apobj.notify(
        body="notification",
        title="my notification ",
    )
```
